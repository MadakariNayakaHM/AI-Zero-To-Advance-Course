## Part C — Reshape, ravel, transpose (§4)

Worked example 4.1: row-major (C) order and the index formula.

```python
t = np.arange(24).reshape(2, 3, 4)
print("shape:", t.shape)
print("t[1, 2, 3] =", t[1, 2, 3], "| formula 1*12 + 2*4 + 3 =", 1*12 + 2*4 + 3)
print("strides (bytes):", t.strides, "itemsize", t.itemsize)
print("reshape(4, -1) ->", t.reshape(4, -1).shape)
print("reshape(-1) ->", t.reshape(-1).shape)
try:
    np.arange(12).reshape(5, -1)
except ValueError as e:
    print("reshape(5, -1) on 12 elements:", e)

a = np.array([1, 2, 3, 4, 5, 6])
print("reshape(3,2):\n", a.reshape(3, 2))
print("reshape(2,-1):\n", a.reshape(2, -1))
print("ravel:", a.reshape(2, 3).ravel())
print("order='F' (column-major):\n", a.reshape(3, 2, order="F"))

m = np.arange(10).reshape(5, 2)
print("m.T shape:", m.T.shape, "| transpose((1,0)) shape:", m.transpose((1, 0)).shape)
img = np.zeros((3, 32, 48))      # channels-first
print("(C,H,W) -> (H,W,C):", img.transpose((1, 2, 0)).shape)
```

## Part D — Indexing and slicing (§5)

Worked example 5.1 uses the array from Assignments 1 and 2.

## Part E — Axes and aggregations (§6)

Worked example 6.1: four students × three subjects.

```python
marks = np.array([[78, 85, 62],
                  [90, 72, 88],
                  [55, 60, 70],
                  [82, 95, 91]])
print("shape:", marks.shape)
print("grand total       :", marks.sum())
print("per student (ax=1):", marks.sum(axis=1))
print("per subject mean  :", marks.mean(axis=0).round(2))
print("subject max       :", marks.max(axis=0))
print("topper index      :", marks.sum(axis=1).argmax())
print("best subject/student (argmax ax=1):", marks.argmax(axis=1))
tot = marks.sum(axis=1, keepdims=True)
print("keepdims shape    :", tot.shape)
print("share of own total (%):\n", (100 * marks / tot).round(1))
print("std per subject (ddof=0):", marks.std(axis=0).round(2))
print("std per subject (ddof=1):", marks.std(axis=0, ddof=1).round(2))
```

```python
# the manual's sum examples, fixed: a 1-D array has only axis 0
a = np.arange(10)
print("a.sum() =", a.sum(), "| a.sum(axis=0) =", a.sum(axis=0), "| keepdims:", a.sum(axis=0, keepdims=True))
try:
    a.sum(axis=1)
except Exception as e:
    print(type(e).__name__, ":", e)
a2 = np.arange(10).reshape(2, 5)
print(a2)
print("axis=0:", a2.sum(axis=0), "| axis=1:", a2.sum(axis=1), "| axis=1 keepdims:", a2.sum(axis=1, keepdims=True).ravel(), a2.sum(axis=1, keepdims=True).shape)
```

## Part F — Joining and splitting (§7)

```python
a = np.ones((3, 5), dtype=np.float32); b = np.zeros((3, 5), dtype=np.float32)
print("concat axis 0:", np.concatenate([a, b]).shape, "| axis 1:", np.concatenate([a, b], axis=1).shape)
x = np.array([1, 2, 3]); y = np.array([4, 5, 6])
print("vstack:\n", np.vstack([x, y]))
print("hstack:", np.hstack([x, y]))
print("stack axis 0:", np.stack([x, y]).shape, "| axis 1:", np.stack([x, y], axis=1).shape)
print("column_stack:\n", np.column_stack([x, y]))
print("split 9 into 3:", np.split(np.arange(9), 3))
print("split at [2, 5]:", np.split(np.arange(9), [2, 5]))
try:
    np.concatenate([np.zeros((2, 3)), np.zeros((4, 3))], axis=1)
except ValueError as e:
    print("axis=1 error:", e)
```

## Part G — Boolean masks, fancy indexing, np.where (§8)

Worked example 8.1: the scores array from Assignment 1.

```python
scores = np.array([45, 82, 67, 33, 91, 58, 74, 29, 88])
mask = scores >= 60
print("mask      :", mask)
print("passed    :", scores[mask], "| count:", mask.sum(), "| fraction:", mask.mean().round(3))
print("positions :", np.where(mask)[0])
print("labels    :", np.where(mask, "Pass", "Fail"))
print("60-80     :", scores[(scores >= 60) & (scores < 80)])
print("fail or >=90:", scores[(scores < 60) | (scores >= 90)])
grades = np.select([scores >= 85, scores >= 70, scores >= 60], ["A", "B", "C"], default="F")
print("np.select :", grades)
print("fancy [0, 4, -1]:", scores[[0, 4, -1]])
print("top 3 (argsort):", scores[np.argsort(scores)[::-1][:3]])
```

```python
# operator precedence trap
try:
    scores[scores >= 60 & scores < 80]
except Exception as e:
    print(type(e).__name__, ":", str(e)[:90])
print("60 & scores =", 60 & scores)
```

## Part H — Vectorization and speed (§9)

Timings depend on the machine; the ratio is what matters.

```python
n = 1_000_000
lst = list(range(n)); arr = np.arange(n)
t0 = time.perf_counter(); s1 = sum(v * v for v in lst); t_loop = time.perf_counter() - t0
t0 = time.perf_counter(); s2 = int((arr * arr).sum()); t_np = time.perf_counter() - t0
print(f"loop {t_loop*1000:.1f} ms | numpy {t_np*1000:.2f} ms | ~{t_loop/t_np:.0f}x faster | equal: {s1 == s2}")
x = np.array([1, 4, 9, 16])
print("sqrt:", np.sqrt(x), "| exp(0,1):", np.exp([0, 1]).round(4), "| log(e):", np.log(np.e))
```

## Part I — Broadcasting (§10)

Worked example 10.1: standardising a feature matrix (height cm, weight kg).

```python
X = np.array([[170, 65],
              [160, 58],
              [180, 80],
              [175, 72]], dtype=float)
mu = X.mean(axis=0); sd = X.std(axis=0)
print("mean:", mu, "| std:", sd.round(4))
Z = (X - mu) / sd
print("Z:\n", Z.round(3))
print("Z mean:", Z.mean(axis=0).round(10), "| Z std:", Z.std(axis=0))
```

```python
def bshape(*shapes):
    try:
        return np.broadcast_shapes(*shapes)
    except ValueError:
        return "error"
for s in [((3, 4), ()), ((3, 4), (4,)), ((3, 4), (3, 1)), ((3, 1), (1, 4)), ((3, 4), (3,)),
          ((2, 1, 3), (4, 1)), ((4, 1), (3,)), ((2, 3), (2,)), ((8, 1, 6, 1), (7, 1, 5))]:
    print(s, "->", bshape(*s))
col = np.array([[10], [20], [30]]); row = np.array([1, 2, 3, 4])
print("col + row:\n", col + row)
```

## Part J — Copy vs view (§11)

```python
a = np.arange(6)
v = a[1:4]; v[0] = 100
print("after editing slice, a =", a, "| shares memory:", np.shares_memory(a, v))
a = np.arange(6)
b = a.reshape(2, 3); b[0, 0] = 99
print("after editing reshape, a =", a)
a = np.arange(6)
c = a[a > 2]; c[0] = -1
print("after editing boolean selection, a =", a, "| shares memory:", np.shares_memory(a, c))
a = np.arange(6)
d = a[1:4].copy(); d[0] = 100
print("after editing .copy(), a =", a)
lst = [0, 1, 2, 3]; s = lst[1:3]; s[0] = 100
print("python list slice is a copy:", lst)
```

## Part K — dtypes and pitfalls (§12)

```python
print(np.array([1, 2, 3]).dtype, np.array([1, 2.5]).dtype, np.array([1, 2.5, True]).dtype, np.array([1, "2"]).dtype)
f = np.array([1.7, -3.9, 2.5])
print("astype(int):", f.astype(int), "| round then int:", np.round(f).astype(int))
u = np.array([200], dtype=np.uint8)
print("uint8 200 + 100 =", u + np.uint8(100), "(wraps mod 256)")
i8 = np.array([127], dtype=np.int8)
print("int8 127 + 1 =", i8 + np.int8(1))
print("0.1 + 0.2 == 0.3:", 0.1 + 0.2 == 0.3, "| isclose:", np.isclose(0.1 + 0.2, 0.3))
print("float64 1000 values:", np.ones(1000).nbytes, "bytes | float32:", np.ones(1000, np.float32).nbytes)
with np.errstate(divide="ignore", invalid="ignore"):
    r = np.array([1.0, 0.0, -1.0]) / 0
print("x / 0:", r, "| isfinite:", np.isfinite(r))
print("nan == nan:", np.nan == np.nan, "| mean with nan:", np.mean([1, np.nan, 3]), "| nanmean:", np.nanmean([1, np.nan, 3]))
```

## Part L — Images as arrays (§13)

Worked example 13.1: a tiny 2×2 RGB image.

```python
I = np.array([[[255, 0, 0], [0, 255, 0]],
              [[0, 0, 255], [255, 255, 255]]], dtype=np.uint8)
print("shape (H, W, C):", I.shape)
print("BGR (channel flip) pixel (0,0):", I[:, :, ::-1][0, 0])
w = np.array([0.299, 0.587, 0.114])
gray = I @ w                       # (2,2,3) @ (3,) -> (2,2)
print("grayscale:\n", gray.round(3))
print("horizontal flip, row 0:", I[:, ::-1][0].tolist())
big = np.arange(6 * 6 * 3).reshape(6, 6, 3)
print("crop 1-pixel border:", big.shape, "->", big[1:-1, 1:-1].shape)
print("brightness +100 clipped:", np.clip(I.astype(int) + 100, 0, 255)[0, 0])
```

## Part M — Random numbers (§14)

```python
rng = np.random.default_rng(42)
print("random(3):", rng.random(3).round(4))
print("integers 1..6:", rng.integers(1, 7, size=5))
rng1 = np.random.default_rng(0); rng2 = np.random.default_rng(0)
print("same seed, same numbers:", np.array_equal(rng1.random(5), rng2.random(5)))
rng = np.random.default_rng(0)
pts = rng.random((100_000, 2))
inside = (pts ** 2).sum(axis=1) <= 1
print("Monte Carlo pi ~", 4 * inside.mean())
rng = np.random.default_rng(1)
idx = rng.permutation(10)
print("permutation:", idx, "| train:", idx[:8], "| test:", idx[8:])
```

## Part N — Linear algebra bridge (§15)

```python
A = np.array([[2, 1], [1, 3]]); bvec = np.array([5, 10])
print("A @ A:\n", A @ A)
print("A * A:\n", A * A)
print("det:", round(np.linalg.det(A), 6), "| solve:", np.linalg.solve(A, bvec))
print("inv:\n", np.linalg.inv(A))
print("norm [3,4]:", np.linalg.norm([3, 4]))
```

## Part O — Case studies (§16)

```python
# 16.1 class result sheet: weighted total with broadcasting
marks = np.array([[78, 85, 62], [90, 72, 88], [55, 60, 70], [82, 95, 91]])
weights = np.array([0.3, 0.3, 0.4])
final = marks @ weights
print("weighted final:", final.round(2))
print("rank order (best first):", np.argsort(-final))
# 16.2 sensor readings: clean bad values with masks
r = np.array([21.5, 22.1, -999.0, 23.0, 85.0, 22.7, np.nan, 21.9])
bad = ~np.isfinite(r) | (r < -50) | (r > 60)
print("bad positions:", np.where(bad)[0])
clean = np.where(bad, np.nan, r)
print("clean mean:", round(np.nanmean(clean), 3), "| naive mean (ignoring nan only):", round(np.nanmean(r), 3))
```
